```python
import numpy as np
import matplotlib.pyplot as plt
import skimage as ski
img_0 = ski.io.imread("evolucion_dominio/0.png")
img = ski.io.imread("evolucion_dominio/7.png")
img_resta =  img - img_0

plt.hist(img.ravel(), bins=128)
plt.yscale("log")
plt.xlabel("valor del píxel")
plt.ylabel("cantidad de píxeles")
plt.show()

plt.figure(figsize=(4, 4))
plt.imshow(img, cmap='gray')
plt.axis('off')
plt.show()

plt.hist(img_resta.ravel(), bins=128)
plt.yscale("log")
plt.xlabel("valor del píxel")
plt.ylabel("cantidad de píxeles")
plt.show()

plt.figure(figsize=(4, 4))
plt.imshow(img_resta, cmap='gray')
plt.axis('off')
plt.show()
```
